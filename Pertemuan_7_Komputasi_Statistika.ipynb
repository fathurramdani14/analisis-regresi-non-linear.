Estimasi Integral, SE dan CI

```python
import numpy as np

rng = np.random.default_rng(42)
a, b, N = 0, 1, 10000
x = rng.uniform(a, b, N)
y = x**2

estimate = (b-a) * y.mean()
se = (b-a) * y.std(ddof=1) / np.sqrt(N)
ci = (estimate - 1.96*se, estimate + 1.96*se)

print("Estimate:", estimate)
print("SE:", se)
print("CI 95%:", ci)
print("Absolute error:", abs(estimate-1/3))
```

Simulasi Berulang dan Coverage

```python
import numpy as np

rng = np.random.default_rng(2026)
R, N = 1000, 100
x = rng.uniform(0, 1, (R, N))
y = x**2
est = y.mean(axis=1)
se = y.std(axis=1, ddof=1) / np.sqrt(N)
lo = est - 1.96*se
hi = est + 1.96*se

coverage = np.mean((lo <= 1/3) & (1/3 <= hi))
print("Coverage empiris:", coverage)
print("SD estimasi:", est.std(ddof=1))
print("SE teoritis:", np.sqrt(4/(45*N)))
```

Probabilitas Normal : Direct vs Indicator



```python
import numpy as np
from scipy.stats import norm

rng = np.random.default_rng(42)
N = 50000
x = rng.uniform(0, 1, N)
z = rng.normal(0, 1, N)

A = norm.pdf(x)
B = ((z > 0) & (z < 1)).astype(float)

for name, y in [("Direct", A), ("Indicator", B)]:
    print(name, "est:", y.mean(),
          "SE:", y.std(ddof=1)/np.sqrt(N))

print("Exact:", norm.cdf(1)-norm.cdf(0))
```

Integral Multidimensi

```python
import numpy as np

rng = np.random.default_rng(42)
N = 50000
xy = rng.uniform(0, 1, (N, 2))
g = np.exp(-np.sum(xy**2, axis=1))

print("Estimate:", g.mean())
print("Estimated SE:", g.std(ddof=1)/np.sqrt(N))
print("Reference: 0.5577462853510335")
```

Bayesian Self-Normalized Importance Sampling

```python
import numpy as np

rng = np.random.default_rng(42)
N = 50000
alpha, beta = 1, 1
success, failure = 7, 3

theta = rng.uniform(0, 1, N)
w = theta**(alpha + success - 1) * \
    (1-theta)**(beta + failure - 1)
mc_mean = np.sum(theta*w) / np.sum(w)
ess = np.sum(w)**2 / np.sum(w**2)
exact = (alpha + success) / (alpha + beta + success + failure)

print("MC posterior mean:", mc_mean)
print("Exact posterior mean:", exact)
print("Weight ESS:", ess)
```
